```python
# Student ID: 242031025
# Algorithm: Breadth First Search (BFS)

import networkx as nx
import matplotlib.pyplot as plt
from collections import deque

# Creating Graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': ['G'],
    'E': ['G'],
    'F': ['G'],
    'G': []
}

# Implementing BFS Algorithm
def bfs(graph, start, goal):

    queue = deque([[start]])
    visited = set()

    while queue:

        path = queue.popleft()
        node = path[-1]

        if node not in visited:

            print("Visiting:", node)
            visited.add(node)

            if node == goal:
                return path

            for neighbor in graph[node]:

                if neighbor not in visited:
                    queue.append(path + [neighbor])

    return None


# Running BFS
start = 'A'
goal = 'G'

path = bfs(graph, start, goal)

print("\nStudent ID: 242031025")
print("Shortest Path:", path)


# Graph Visualization
G = nx.DiGraph()

for node in graph:
    for neighbor in graph[node]:
        G.add_edge(node, neighbor)

# Node Positions
pos = {
    'A': (0, 2),
    'B': (-1, 1),
    'C': (1, 1),
    'D': (-1.5, 0),
    'E': (-0.5, 0),
    'F': (1, 0),
    'G': (0, -1)
}

plt.figure(figsize=(9, 6))

# Draw Graph
nx.draw(
    G, pos,
    with_labels=True,
    node_color='lightblue',
    node_size=1800,
    font_size=12,
    arrows=True
)

# Highlight Shortest Path
if path:

    path_edges = list(zip(path, path[1:]))

    nx.draw_networkx_nodes(
        G, pos,
        nodelist=path,
        node_color='lightgreen',
        node_size=1800
    )

    nx.draw_networkx_edges(
        G, pos,
        edgelist=path_edges,
        edge_color='red',
        width=3,
        arrows=True
    )

plt.title("BFS Visualization - Student ID: 242031025")
plt.axis('off')
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt

graph = {
    'B': ['U', 'S'],
    'U': ['H', 'R'],
    'S': ['A', 'T'],
    'H': [],
    'R': [],
    'A': [],
    'T': []
}

def DFS(graph, start):
    visited = []
    stack = [start]

    while stack:
        node = stack.pop()

        if node not in visited:
            visited.append(node)

            for neighbor in reversed(graph[node]):
                if neighbor not in visited:
                    stack.append(neighbor)

    return visited

result = DFS(graph, 'B')
print("DFS Traversal:", result)

G = nx.Graph(graph)
pos = nx.spring_layout(G, seed=10)

plt.ion()
fig, ax = plt.subplots()

visited = []

for node in result:
    visited.append(node)
    ax.clear()

    colors = [
        'orange' if n == node else
        'lightgreen' if n in visited else
        'skyblue' for n in G.nodes()
    ]

    nx.draw(G, pos, with_labels=True,
            node_color=colors, node_size=2000,
            font_size=14, ax=ax)

    ax.set_title("DFS - BUSHRAT | Visiting: " + node)
    plt.pause(1)

plt.ioff()
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt

graph = {
    '2': ['4', '0'],
    '4': ['3', '1'],
    '0': ['5'],
    '3': [],
    '1': [],
    '5': []
}

def DLS(graph, node, target, limit, visited):
    visited.append(node)

    if node == target:
        return True

    if limit <= 0:
        return False

    for neighbor in graph[node]:
        if neighbor not in visited:
            if DLS(graph, neighbor, target,
                   limit - 1, visited):
                return True

    return False

visited = []
found = DLS(graph, '2', '5', 2, visited)

print("DLS Traversal:", visited)
print("Target Found:", found)

G = nx.Graph(graph)
pos = nx.spring_layout(G, seed=10)

plt.ion()
fig, ax = plt.subplots()

done = []

for node in visited:
    done.append(node)
    ax.clear()

    colors = [
        'orange' if n == node else
        'lightgreen' if n in done else
        'skyblue' for n in G.nodes()
    ]

    nx.draw(G, pos, with_labels=True,
            node_color=colors, node_size=2000,
            font_size=14, ax=ax)

    ax.set_title("DLS - ID 242031025 | Visiting: " + node)
    plt.pause(1)

plt.ioff()
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt

graph = {
    'T': ['A', 'S'],
    'A': ['R', 'I'],
    'S': ['N'],
    'R': [],
    'I': [],
    'N': []
}

def IDDFS(start, target, max_depth):

    G = nx.Graph()

    for node, neighbors in graph.items():
        for neighbor in neighbors:
            G.add_edge(node, neighbor)

    pos = {
        'T': (0, 3),
        'A': (-1, 2),
        'S': (1, 2),
        'R': (-1.5, 1),
        'I': (-0.5, 1),
        'N': (1, 1)
    }

    for depth in range(max_depth + 1):

        visited = []

        def DLS(node, level):
            visited.append(node)

            if node == target:
                return True

            if level == depth:
                return False

            for neighbor in graph[node]:
                if DLS(neighbor, level + 1):
                    return True

            return False

        found = DLS(start, 0)

        print("Depth:", depth)
        print("Visited:", visited)

        plt.figure(figsize=(8, 5))

        colors = [
            'orange' if n == target and found and n in visited
            else 'lightgreen' if n in visited
            else 'skyblue'
            for n in G.nodes()
        ]

        nx.draw(
            G, pos,
            with_labels=True,
            node_color=colors,
            node_size=2200,
            font_size=15,
            font_weight='bold'
        )

        plt.title(f"IDDFS - TASRIN | Depth Limit: {depth}")
        plt.show()

        if found:
            break

IDDFS('T', 'N', 3)
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import heapq

graph = {
    '2': [('4', 2), ('0', 4)],
    '4': [('3', 1), ('1', 5)],
    '0': [('5', 2)],
    '3': [('5', 3)],
    '1': [('5', 2)],
    '5': []
}

def UCS(start, target):

    pq = [(0, start, [start])]
    visited = []
    steps = []

    while pq:

        cost, node, path = heapq.heappop(pq)

        if node in visited:
            continue

        visited.append(node)
        steps.append((node, visited.copy(), cost))

        if node == target:
            print("Path:", path)
            print("Total Cost:", cost)
            break

        for neighbor, weight in graph[node]:

            if neighbor not in visited:
                heapq.heappush(
                    pq,
                    (cost + weight, neighbor, path + [neighbor])
                )

    G = nx.Graph()

    for node, neighbors in graph.items():
        for neighbor, weight in neighbors:
            G.add_edge(node, neighbor, weight=weight)

    pos = {
        '2': (0, 3),
        '4': (-1, 2),
        '0': (1, 2),
        '3': (-1.5, 1),
        '1': (-0.5, 1),
        '5': (1, 1)
    }

    for node, done, cost in steps:

        plt.figure(figsize=(8, 5))

        colors = [
            'orange' if n == node
            else 'lightgreen' if n in done
            else 'skyblue'
            for n in G.nodes()
        ]

        nx.draw(
            G, pos,
            with_labels=True,
            node_color=colors,
            node_size=2200,
            font_size=15,
            font_weight='bold'
        )

        nx.draw_networkx_edge_labels(
            G, pos,
            edge_labels=nx.get_edge_attributes(G, 'weight')
        )

        plt.title(f"UCS - ID 242031025 | Visiting: {node} | Cost: {cost}")
        plt.show()

UCS('2', '5')
```
